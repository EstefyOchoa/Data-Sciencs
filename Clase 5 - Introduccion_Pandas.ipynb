## MASCARAS BOOLEANAS

```python
# df['col1'] > 10     --> Esto es una mascara booleana ya que devuelve valores booleanos.
# df[df['col1] > 10]  --> Solo trae datos de aquellas filas que cumplas con las mascara booleana.

# Traer solo las filas donde 'math score' sea mayor a 7.0
students_df['math score'] > 7.0 # Esta es la MASCARA BOOLEANA

students_df[students_df['math score'] > 7.0] # Filtramos los datos que cumplen las condicion
```

```python
# Traer solo las filas donde reading score sea menor a 60
readingScore_mask = students_df['reading score'] > 60 # Esta es la mascara booleana
students_df[readingScore_mask] # Filtramos los datos que cumplen las condicion
```

```python
# Traer solo las filas donde gender sea igual a female
gender_mask = students_df['gender'] == 'female'  # Esta es la mascara booleana
students_df[gender_mask] # Filtramos los datos que cumplen las condicion
```

```python
# Traer solo aquellas filas donde lunch sea distinto a standard
lunch_mask = students_df['lunch'] != 'standard'  # Esta es la mascara booleana
students_df[lunch_mask]  # Filtramos los datos que cumplen las condicion
```

```python
# Traer aquellos estudiantes que posean reading and writing score mayor a 70

mask = (students_df['reading score'] > 70) & (students_df['writing score'] > 70)
students_df[mask]
```
