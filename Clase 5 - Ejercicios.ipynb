```python
# Selecciona el maximo de datos faltantes. Calcula que porcentaje de los datos representan los datos faltantes.
max_nan = expvida.isna().sum().max()
```

```python
# Obtengo la cantidad de filas total y divido el maximo valor de datos faltantes por este numero y lo multiplico por 100
total_row = expvida.index.size

print(f"El porcentaje maximo de filas con datos faltantes es: {(max_nan/total_row)*100:.2f}%")
```

```python
porcentaje = (max_nan/total_row)*100
porcentaje_str = str(porcentaje) # Convierto a string
print(f"{porcentaje}")
print(f"{porcentaje_str}")
print(f"{type(porcentaje)}")
print(f"{type(porcentaje_str)}")
```

```python
expvida.columns = expvida.columns.str.strip() # Elimino los espacios que hay en el nombre de cada columna. para luego poder renombrar.
expvida
```

```python
# Renombra la columna 'Life expectancy ' como 'life_expectancy', ' BMI ' como 'bmi' y ''Measles ' como 'measles'. 
# No olvides de usar `inplace` para modificar el DataFrame original. Luego chequea que el cambio se haya hecho correctamente.

expvida = expvida.rename(columns={"Life expectancy":"life_expectancy","BMI":"bmi","Measles":"measles"})
expvida
```

```python
# ¿Cual es el promedio de expectativa de vida (Life expectancy) en general?
print(f"El promedio genera es: {expvida['life_expectancy'].mean():.2f}")
```

```python
# ¿Cual es el promedio de expectativa de vida (Life expectancy) por país?
expvida.groupby('Country')['life_expectancy'].mean()
```

```python
# ¿Cual es la expectativa de vida mas alta y mas baja? Usa la funcion de pandas y la funcion de NumPy y compara el resultado. 
# ¿Que paises tienen esas expectativas de vida?

expectancy_numpy = np.array(expvida['life_expectancy'])
# Numpy no ignora los Nan (si los hay), por eso me muestra como resultado un nan para las expectativas mas altas y mas bajas.

expectancy_max_np = np.nanmax(expectancy_numpy) # Retoran los valores max ignorando los nan
expectancy_min_np = np.nanmin(expectancy_numpy) # Idem para los min.

expectancy_max_pd = expvida['life_expectancy'].max()
expectancy_min_pd = expvida['life_expectancy'].min()

print(f"Por NUMPY: \nExpectativas mas altas: {expectancy_max_np} \nExpectativas mas bajas: {expectancy_min_np}")
print(f"Por PANDAS: \nExpectativas mas altas: {expectancy_max_pd} \nExpectativas mas bajas: {expectancy_min_pd}")
```

```python
# # Filtro las columnas para mostrar solo las filas que tengan valor igual al maximo
expectancy_row_max = expvida['life_expectancy'] == expectancy_max_pd
expvida[expectancy_row_max]
```

```python
# Filtro las columnas para mostrar solo las filas que tengan valor igual al minimo
expectancy_row_min = expvida['life_expectancy'] == expectancy_min_pd
expvida.loc[expectancy_row_min]
```

```python
# ¿Cuantas categorias encontramos en la columna "Status"?
expvida['Status'].unique()
```

```python
# ¿Cuantos paises se ubican en cada una de esas categorias de "Status"? _**Pista**_: Lee acerca de la función
category = expvida.groupby('Status')['Country'].value_counts()
category
```
